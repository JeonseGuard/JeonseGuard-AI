```python
# 데이터 불러오기
import pandas as pd

file_path = "/content/drive/MyDrive/JeonseGuard/실거래가/전월세/오피스텔/202501_오피스텔_전세_실거래가.csv" # CSV 파일 경로 지정
df = pd.read_csv(file_path, encoding='cp949') # CP949 인코딩
```

```python
# 상위 5개 확인
df.head()
```

```python
# 컬럼명 확인
print(df.columns)
```

```python
# 특정 컬럼 값만 확인
df["보증금(만원)"].head()
```

```python
# 전세만 필터링
jeonse_df = df[df["전월세구분"] == "전세"]
jeonse_df.head()
```

```python
# 주택유형 컬럼 추가
df["주택유형"] = "오피스텔"
```

```python
# 변경 매핑 딕셔너리 정의
renamed_columns = {
    "시군구": "address",
    "본번": "bun",
    "부번": "ji",
    "층": "floor",
    "전월세구분": "rent_type",
    "계약년월": "contract_year_month",
    "보증금(만원)": "price",
    "주택유형": "housing_type"
}
```

```python
# 매핑에 해당하는 컬럼만 필터링
df = df.rename(columns=renamed_columns)
changed_df = df[list(renamed_columns.values())]
changed_df.head()
```

```python
# rent_type 컬럼에서 "전세"만 필터링
jeonse_df = changed_df[changed_df["rent_type"] == "전세"].reset_index(drop=True)
jeonse_df.head()
```

```python
# 쉼표 제거 및 문자열을 정수로 변환
jeonse_df["price"] = (
    jeonse_df["price"]
    .astype(str) # 문자열로 변환 (안전)
    .str.replace(",", "") # 쉼표 제거
    .astype(int) # 정수형으로 변환
    * 10000 # 만원 → 원 변환
)
```

```python
# 변환된 price 컬럼 확인
jeonse_df[["price"]].head()
```

```python
# price 컬럼의 값을 쉼표가 포함된 문자열로 변환
jeonse_df["price"] = jeonse_df["price"].apply(lambda x: format(x, ","))
jeonse_df.head()
```

```python
# 결측치 확인
missing_counts = jeonse_df.isnull().sum()
print("📌 결측치가 있는 컬럼: ", missing_counts[missing_counts > 0])
```

```python
# 전체 중복된 행의 수 확인
duplicate_count = jeonse_df.duplicated().sum()
print(f"📌 중복된 행의 수: {duplicate_count}")
```

```python
# 중복 제거 (기본: 모든 열 기준, keep='first')
jeonse_df = jeonse_df.drop_duplicates().reset_index(drop=True)
```

```python
# INSERT 구문 생성
insert_header = """INSERT INTO jeonse_transaction_officetel (address, bun, ji, floor, rent_type, contract_year_month, price, housing_type, created_at, updated_at)
VALUES """
```

```python
# 행별로 SQL 값 문자열 생성
values = []
```

```python
# 각 행을 values 리스트에 추가
for _, row in jeonse_df.iterrows():
    values.append(f"('{row['address']}', '{row['bun']}', '{row['ji']}', '{row['floor']}', '{row['rent_type']}', '{row['contract_year_month']}', '{row['price']}', '{row['housing_type']}', NOW(), NOW())")
```

```python
# INSERT 구문 상위 5개만 출력
preview_sql = insert_header + ",\n       ".join(values[:5]) + ";"
print(preview_sql)
```

```python
# INSERT 구문 조립
insert_sql = insert_header + ",\n       ".join(values) + ";"
```

```python
# 파일 저장
file_name = "V25__insert_jeonse_transaction_officetel_202501.sql"

with open(file_name, "w", encoding="utf-8") as f:
    f.write(insert_sql)

print(f"{file_name} 파일이 생성되었습니다.")
```
